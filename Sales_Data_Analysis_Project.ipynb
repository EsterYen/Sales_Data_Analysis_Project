```python
import pandas as pd

sales_data = {
    "Date": [
        "2025-01-01",
        "2025-01-05",
        "2025-01-10",
        "2025-01-15",
        "2025-01-20",
        "2025-01-25"
    ],
    "Product": [
        "Laptop",
        "Phone",
        "Tablet",
        "Laptop",
        "Monitor",
        "Phone"
    ],
    "Region": [
        "West",
        "East",
        "West",
        "North",
        "South",
        "East"
    ],
    "Sales": [
        1200,
        850,
        650,
        1500,
        400,
        900
    ],
    "Quantity": [
        12,
        20,
        15,
        15,
        8,
        18
    ]
}

df = pd.DataFrame(sales_data)

df
```

```python
df.info()
```

```python
df["Date"] = pd.to_datetime(df["Date"])

df.info()
```

```python
df.groupby("Product")["Sales"].sum()
```

```python
product_sales = df.groupby("Product")["Sales"].sum()

product_sales.sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

product_sales.sort_values().plot(kind="barh")

plt.xlabel("Sales")
plt.ylabel("Product")
plt.title("Sales by Product")

plt.show()
```

```python
product_sales = df.groupby("Product")["Sales"].sum()

product_sales
```

```python
import matplotlib.pyplot as plt

product_sales.sort_values().plot(kind="barh", figsize=(8,4))

plt.xlabel("Sales")
plt.ylabel("Product")
plt.title("Total Sales by Product")

plt.show()
```

```python
ax = product_sales.sort_values().plot(
    kind="barh",
    figsize=(8,4)
)

for i, value in enumerate(product_sales.sort_values()):
    ax.text(value, i, str(value))

plt.xlabel("Sales")
plt.ylabel("Product")
plt.title("Total Sales by Product")

plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum()
```

```python
import matplotlib.pyplot as plt

region_sales.sort_values().plot(
    kind="barh",
    figsize=(8,4)
)

plt.xlabel("Sales")
plt.ylabel("Region")
plt.title("Total Sales by Region")

plt.show()
```

```python
sorted_region_sales = region_sales.sort_values()

ax = sorted_region_sales.plot(
    kind="barh",
    figsize=(8,4)
)

for i, value in enumerate(sorted_region_sales):
    ax.text(value, i, str(value))

plt.xlabel("Sales")
plt.ylabel("Region")
plt.title("Total Sales by Region")

plt.show()
```
