```python
# Calculate total sales by product
sales_by_product = df.groupby("Product")["Sales"].sum()

# Create the chart
plt.figure(figsize=(8, 5))
bars = plt.bar(sales_by_product.index, sales_by_product.values)

# Add title and labels
plt.title("Total Sales by Product", fontsize=16)
plt.xlabel("Product", fontsize=12)
plt.ylabel("Total Sales ($)", fontsize=12)

# Add value labels on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f"{height}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

sales_data = {
    "Date": [
        "2025-01-01",
        "2025-01-05",
        "2025-01-10",
        "2025-01-15",
        "2025-01-20",
        "2025-01-25"
    ],
    "Product": [
        "Laptop",
        "Phone",
        "Tablet",
        "Laptop",
        "Monitor",
        "Phone"
    ],
    "Region": [
        "West",
        "East",
        "West",
        "North",
        "South",
        "East"
    ],
    "Sales": [
        1200,
        850,
        650,
        1500,
        400,
        900
    ],
    "Quantity": [
        12,
        20,
        15,
        15,
        8,
        18
    ]
}

df = pd.DataFrame(sales_data)

df
```

```python
product_sales = df.groupby("Product")["Sales"].sum()

product_sales.sort_values(ascending=False)
```

```python
df.groupby("Product")["Sales"].sum()
```

```python
id="x2v4p9"
df = df.sort_values("Date")

df
```

```python
# Sort data chronologically
sales_trend = df.sort_values("Date")

plt.figure(figsize=(9, 4))

plt.plot(
    sales_trend["Date"],
    sales_trend["Sales"],
    marker="o"
)

plt.title(
    "Sales Performance Trend Over Time",
    fontsize=16
)

plt.xlabel("Date")
plt.ylabel("Sales ($)")

plt.xticks(rotation=45)

plt.grid(True)

plt.tight_layout()

plt.show()
```

```python
product_sales = df.groupby("Product")["Sales"].sum()

product_sales
```

```python
region_sales = df.groupby("Region")["Sales"].sum()
```

```python
# Sort sales by region (highest first)
region_sales_sorted = region_sales.sort_values(ascending=True)

# Create horizontal bar chart
plt.figure(figsize=(8, 4))

bars = plt.barh(
    region_sales_sorted.index,
    region_sales_sorted.values
)

# Title and labels
plt.title("Regional Sales Performance", fontsize=16)
plt.xlabel("Total Sales ($)", fontsize=12)
plt.ylabel("Region", fontsize=12)

# Add value labels
for bar in bars:
    width = bar.get_width()
    plt.text(
        width,
        bar.get_y() + bar.get_height()/2,
        f"{width}",
        va="center"
    )

plt.tight_layout()
plt.show()
```
